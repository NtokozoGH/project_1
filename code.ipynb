```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sklearn
```

```python
df = pd.read_csv("my_spending.csv")
df
```

```python
# 1. Total amount spent
total_spent = df["amount"].sum()
```

```python
# 2. Number of transactions
total_transactions = len(df)
```

```python
# 3. Average amount spent
average_spent = df["amount"].mean()
```

```python
print("Total spent:", total_spent)
print("Number of transactions:", total_transactions)
print("Average spent:", round(average_spent, 2))
```

```python
# Convert date column to datetime
df["date"] = pd.to_datetime(df["date"])
```

```python
# 4. Total spent in September 2026
september_spent = df[
    (df["date"].dt.year == 2026) &
    (df["date"].dt.month == 9)
]["amount"].sum()

print("4. Total spent in September 2026:", september_spent)
```

```python
# 5. Single day with the highest total spend
daily_spending = df.groupby("date")["amount"].sum()

highest_spend_day = daily_spending.idxmax()
highest_spend_amount = daily_spending.max()

print("5. Highest spending day:", highest_spend_day)
print("   Amount spent:", highest_spend_amount)
```

```python
# 6. Average daily spend for the month
average_daily_spend = september_spent / 30

print("6. Average daily spend:", round(average_daily_spend, 2))
```

```python
# 7. Category with the highest total spend
category_spend = df.groupby("category")["amount"].sum()

highest_category = category_spend.idxmax()
highest_amount = category_spend.max()

print("7. Highest spending category:", highest_category)
print("   Total spent:", highest_amount)
```

```python
# 8. Category with the most transactions
transaction_count = df["category"].value_counts()

most_transactions = transaction_count.idxmax()
most_transaction_count = transaction_count.max()

print("8. Category with most transactions:", most_transactions)
print("   Number of transactions:", most_transaction_count)
```

```python
# 9. Percentage of total spend that is Groceries
groceries_spend = df[df["category"] == "Groceries"]["amount"].sum()
total_spend = df["amount"].sum()

groceries_percentage = (groceries_spend / total_spend) * 100

print("9. Groceries percentage of total spend:",
      round(groceries_percentage, 2), "%")
```

```python
# 10. Categories sorted from highest to lowest spend
sorted_categories = category_spend.sort_values(ascending=False)

print("10. Categories from highest to lowest spend:")
print(sorted_categories)
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
largest_transaction = df.loc[df["amount"].idxmax()]

print("Amount:", largest_transaction["amount"])
print("Category:", largest_transaction["category"])
print("Date:", largest_transaction["date"].date())
```

```python
above_300 = df[df["amount"] > 300]

print(above_300)
```

```python
weekly_spend = df.groupby(
    df["date"].dt.to_period("W-SUN")
)["amount"].sum()

print(weekly_spend)

most_expensive_week = weekly_spend.idxmax()
most_expensive_amount = weekly_spend.max()

print("\nMost expensive week:", most_expensive_week)
print("Amount spent:", most_expensive_amount)
```

```python
budget = 3500

total_spend = df["amount"].sum()
amount_over_budget = total_spend - budget

print("Total spend:", total_spend)
print("Over budget by:", amount_over_budget)
```

```python
category_spend = df.groupby("category")["amount"].sum()

print(category_spend.sort_values(ascending=False))
```
